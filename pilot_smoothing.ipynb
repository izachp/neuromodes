```python
import numpy as np
from neuromodes.io import fetch_surf
from neuromodes import EigenSolver
from neuromodes.mesh import smooth
from nsbutils.plotting_pyvista import plot_surf
```

```python
geometry, mask = fetch_surf()
solver = EigenSolver(geometry, mask).compute_lbo()
```

```python
dts = [10**x for x in range(-1, 4)]
data = np.empty((solver.n_verts, 1+len(dts)))
data[:, 0] = np.random.default_rng(0).standard_normal(size=solver.n_verts)
for i, dt in enumerate(dts):
    data[:, i+1] = smooth(data[:, 0], solver.mass, solver.stiffness, dt=dt)
clim = (data.min(), data.max())
```

```python
fig = plot_surf(surf={'lh': solver.geometry}, data={'lh': data[:, 3]}, cmap='seismic', clim=clim)
fig.show()
```
