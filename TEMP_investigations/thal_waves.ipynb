```python
from neuromodes import EigenSolver
from neuromodes.io import fetch_vol
from neuromodes.TEMP_utils import plot_mesh_data
```

```python
dt = 0.1
```

```python
thal = fetch_vol('thalamus')
solver = EigenSolver(thal).solve(500, seed=0)
```

```python
neural = solver.simulate_waves(nt=1000, dt=dt)
```

```python
plot_mesh_data(solver.geometry, neural[:, ::10], cnorm=True, plot_edges=True)
```

```python
bold = solver.bold_transform(neural, dt=dt)
```

```python
plot_mesh_data(solver.geometry, bold[:, ::10], cnorm=True, plot_edges=True)
```
