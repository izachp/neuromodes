```python
import numpy as np
```

### Balloon model padding

```python
kappa = 0.65
tau = 0.98
alpha = 0.32
dt = 0.01
padding_tol = 1e-3
```

```python
decay_eff = np.min((kappa / 2, 1 / tau, 1 / (alpha * tau)))
t_pad = -np.log(padding_tol) / decay_eff
n_pad = int(np.ceil(t_pad / dt))
print(f"Decay effective: {decay_eff:.4f}, t_pad: {t_pad:.4f}, n_pad: {n_pad}")
```
